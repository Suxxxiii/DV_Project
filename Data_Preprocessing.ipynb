```python
import json
from pathlib import Path

Source_Path = Path.cwd() / "Dataset" / "COROPplus_52_enriched_2024.geojson"
with Source_Path.open(encoding="utf-8-sig") as Source_File:
    Regions = json.load(Source_File)["features"]
```

```python
print("Region count:", len(Regions))
for Region in Regions:
    print( Region["properties"]["statnaam"])
```

```python
Property_Names = list(Regions[0]["properties"])
print("\nProperty count:", len(Property_Names))
print("Property | Data types | Empty | Type mismatch | Min | Max")

# Check empty values and compare types across regions.
for Property_Name in Property_Names:
    Data_Types = set()
    Numbers = []
    Empty_Count = 0
    for Region in Regions:
        Value = Region["properties"].get(Property_Name)
        if Value is None or isinstance(Value, str) and not Value.strip():
            Empty_Count += 1
        elif type(Value) in (int, float):
            Data_Types.add("number")
            Numbers.append(Value)
        else:
            Data_Types.add("string" if isinstance(Value, str) else type(Value).__name__)

    Min_Value = min(Numbers) if Numbers else "-"
    Max_Value = max(Numbers) if Numbers else "-"
    Type_Mismatch = "Yes" if len(Data_Types) > 1 else "No"
    print(f"{Property_Name} | {', '.join(sorted(Data_Types))} | {Empty_Count} | {Type_Mismatch} | {Min_Value} | {Max_Value}")
```
